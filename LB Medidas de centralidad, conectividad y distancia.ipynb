# Medidas de centralidad

A través de las medidas de centralidad buscamos identificar a los nodos que tengan la mayor importancia dentro de la red.

En esta libreta exploraremos algunas medidas de centralidad clásicas.

# Grado

El grado de un nodo es el número de conexiones que tiene un nodo.

```python
nx.degree(g)
```

```python
import operator

dg = nx.degree(g_r)
dg_s = sorted(dg, key=operator.itemgetter(1), reverse=True)
dg_s[:20]
```

```python
# elimina nodos con pocas conexiones

def trim_degrees(g, degree=1):
    g2=g.copy()
    d=nx.degree(g2)
    for n in g.nodes():
        if d[n]<=degree:
            g2.remove_node(n)
    return g2
```

```python
g_r2 = trim_degrees(g_r,10)
len(g_r2)
```

# Cercanía

Definimos la **cercanía** para un nodo como el inverso de la distancia promedio al resto de los nodos:

$$C(i) = \frac{n-1}{\sum_{j\neq i}d_{i,j}}$$

siendo $d_{ij}$ la distancia entre los nodos $i$ y $j$,

```python
nx.closeness_centrality(g)
```

```python
nx.draw_networkx(g)
plt.show()
```

```python
dc = nx.closeness_centrality(g_r2)
dc_srtd = sorted(dc, key=operator.itemgetter(1), reverse=True)
dc_srtd[:20]
```

## Intermediación

Permite identificar *cuellos de botella*. La medida de intermediación se define como:

$$B(i) = \sum_{k,j} \frac{\text{I}[r_{k,j}(i)]}{\text{I}[r_{k,j}]}$$

* $\text{I}[r_{k,j}(i)]$ indica si una ruta óptima entre $k$ y $j$ pasa por el nodo $i$.
* $\text{I}[r_{k,j}]$ indica si existe una ruta entre $k$ y $j$.


```python
nx.betweenness_centrality(g)
```

# Medidas de conectividad

Permiten evaluar la robustez de las conexiones entre los elementos de una gráfica y, principalmente, encontrar comunidades entre los elementos de la red.

## Coeficiente de clusterización

$$lc_i = \frac{\text{total de parejas de vecinos de i relacionados}}{\text{total de parejas de vecinos de i}}$$

(los amigos de $i$ son también amigos entre sí)

```python
nx.clustering(g, 7)
```

Identificamos el nodo con el mayor coeficiente de clusterización:

```python
cc_m = 0
nm = None
for n in g:
    cc = nx.clustering(g, n)
    if cc > cc_m:
        cc_m = cc
        nm = n

print(nm)
print(cc_m)
```

```python
nx.draw_networkx(g)
plt.show()
```

```python
# promedio del coeficiente de clusterización
nx.average_clustering(g)
```

## Transitividad

Proporción de triadas abiertas:

$$t_g = \frac{\text{total de triadas abiertas}}{\text{total de triadas cerradas}}$$

Indica la probabilidad de que dos nodos con un amigo común también sean amigos.


```python
nx.transitivity(g)
```

```python
nx.draw_networkx(g)
plt.show()
```

## Clique

Los cliques representan comunidades en una red scial.

Un clique es un conjunto de nodos completamente conectados (cualquier pareja del conjunto debe estar conectada en la gráfica)

## Clique máximo

Un clique que deja de ser clique al agregar cualquier otro nodo

```python
cliqs =  nx.find_cliques(g)

for c in cliqs:
    print(c)
```
