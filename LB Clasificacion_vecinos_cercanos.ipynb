2. **Problemas de clasificación**: si la variable de respuesta es categórica.

# Clasificación


Como se acaba de mencionar, disponemos de un conjunto de datos con un conjunto de predictores, $X_i$ ($i = 1, 2, \ldots, d$), y una variable de respuesta categórica $Y \in \{l_1, l_2, \ldots, l_K\}$

\begin{matrix}
    \hline
    \mathbf{X_1} & \mathbf{X_2} & \cdots & \mathbf{X_d} & \mathbf{Y} \\
    \hline    
    x_{11} & x_{12} & \cdots & x_{1d} & y_1 \\
    x_{21} & x_{22} & \cdots & x_{2d} & y_2 \\
    \vdots & \vdots & \cdots & \vdots & \vdots \\
    x_{n1} & x_{n2} & \cdots & x_{nd} & y_n \\
    \hline
\end{matrix}

Buscamos una función, $f$, que asigne la etiqueta correcta a cada vector de predictores:

$$\color{blue}{f(x_1, x_2, \ldots, x_d) = y}$$

Las siguientes celdas muestran algunos conjuntos de datos con esta característica.

## Vecinos más cercanos

Posiblemente, la metodología de clasificación más intuitiva es la de **vecinos más cercanos**.

```r
virginica <- subset(iris, Species=="virginica")
plot(virginica$Petal.Length, virginica$Petal.Width, col="blue", pch=16, xlim=c(1,7), ylim=c(0,2.7), xlab = "Longitud del pétalo", ylab = "Anchura del pétalo")

setosa <- subset(iris, Species=="setosa")
points(setosa$Petal.Length, setosa$Petal.Width, col="red", pch=16)

versicolor <- subset(iris, Species=="versicolor")
points(versicolor$Petal.Length, versicolor$Petal.Width, col="green", pch=16)
legend(2, 2.5, legend=c("virginica", "setosa", "versicolor"), col=c("blue", "red", "green"), pch=16)
```

Tenemos una **nueva observación** con las siguientes características:

* Longitud del pétalo: 4.7 cm
* Anchura del pétalo: 1.5 cm

¿A qué clase debe asignarse?

```r
new_obs <- data.frame(Petal.Length=4.7, Petal.Width=1.5)

plot(virginica$Petal.Length, virginica$Petal.Width, col="blue", pch=16, xlim=c(1,7), ylim=c(0,2.7), main="¿Qué etiqueta debe asignarse?", xlab = "Longitud del pétalo", ylab = "Anchura del pétalo")
points(setosa$Petal.Length, setosa$Petal.Width, col="red", pch=16)
points(versicolor$Petal.Length, versicolor$Petal.Width, col="green", pch=16)
points(new_obs$Petal.Length,new_obs$Petal.Width, col="black",pch=16, cex=1.5)

legend(2, 2.5, legend=c("virginica", "setosa", "versicolor"), col=c("blue", "red", "green"), pch=16)
```

Para asignar etiqueta a la nueva observación, identificamos a la observación en el conjunto de referencia que esté más cercana a la nueva.

Para ese fin, evaluamos la distancia entre la nueva observación y cada una de las observaciones del conjunto de referencia.

```r
dat <- iris

dat$dx <- dat$Petal.Length-new_obs$Petal.Length
dat$dy <- dat$Petal.Width-new_obs$Petal.Width

dat$dist <- sqrt(dat$dx**2+dat$dy**2)
dat[order(dat$dist),][1:5,]
```

Observamos que la más cercana es de la especie *versicolor* y asignamos entonces esa etiqueta al nuevo caso.

En el ejemplo anterior ha sido muy claro indentificar la clase de la nueva observación, **¿qué clase asignarías en el siguiente caso?**

```r
new_obs <- data.frame(Petal.Length=5.1, Petal.Width=1.55)

plot(virginica$Petal.Length, virginica$Petal.Width, col="blue", pch=16, xlim=c(1,7), ylim=c(0,2.7), main="¿Qué etiqueta debe asignarse?", xlab = "Longitud del pétalo", ylab = "Anchura del pétalo")
points(setosa$Petal.Length, setosa$Petal.Width, col="red", pch=16)
points(versicolor$Petal.Length, versicolor$Petal.Width, col="green", pch=16)
points(new_obs$Petal.Length,new_obs$Petal.Width, col="black",pch=16, cex=1.5)

legend(2, 2.5, legend=c("virginica", "setosa", "versicolor"), col=c("blue", "red", "green"), pch=16)
```

En lugar de considerar al **vecino más cercano**, podríamos basarnos en los primeros 2, 3, 4, ...

```r
dat$dx <- dat$Petal.Length-new_obs$Petal.Length
dat$dy <- dat$Petal.Width-new_obs$Petal.Width

dat$dist <- sqrt(dat$dx**2+dat$dy**2)
dat_ordrd <- dat[order(dat$dist),]
dat_ordrd[1:9,]
```

Esta metodología para clasificar datos se denomina **<font color="red">vecinos más cercanos</font>**

Desde $\texttt{R}$ podemos aplicar el método $\texttt{knn}$

```r
#install.packages("class")
library(class)
```

```r
# observa que consideramos a todos los predictores
X <- dat[,c("Sepal.Length", "Sepal.Width", "Petal.Length", "Petal.Width")]
# variable de respuesta
Y <- dat$Species
```

Para evaluar la calidad del *clasificador* separamos un conjunto de **datos de validación**

```r
set.seed(100)

n <- nrow(dat)
train <- sample(1:n,0.75*n)

X_train <- X[train,]
Y_train <- Y[train]
X_val <- X[-train,]
Y_val <- Y[-train]

#construimmos el "modelo" con los datos de entrenamiento y validamos con el conjunto de prueba
yp_val <- knn(X_train, X_val, Y_train, k=1)

df <- data.frame(y=Y_val, yp=yp_val)
df
```

```r
# medimos la exactitud
sum(df$y==df$yp)/nrow(df)
```

**En este ejemplo empleamos k = 1**, pero en general,
¿cómo escoger el parámetro $k$?

* Podemos evaluar diferentes valores y escoger el que brinde la mayor exactitud sobre el conjunto de validación:

```r
accs <- c()

for(k in 1:10){
    yp_val <- knn(X_train, X_val, Y_train, k=k)
    acc <- sum(Y_val==yp_val)/length(yp_val) #observa que evaluamos sobre el conjunto de prueba (que en este caso se debe llamar conjunto de validación)
    print(paste0(k, " ---> ",acc))
    accs <- c(accs,acc)
}

plot(1:10, accs, type = "l")
```

**¿Qué efecto tiene el parámetro $k$ en las características del clasificador?**
