## Descripción del proyecto

Este notebook presenta una práctica de análisis de complejidad algorítmica utilizando la notación **Big-O**, con el objetivo de comprender cómo aumenta el número de operaciones de un algoritmo conforme crece el volumen de datos procesados.

Durante el ejercicio se estudian diferentes estructuras de programación, desde operaciones de acceso directo y recorridos simples hasta ciclos anidados, búsqueda binaria, algoritmos de ordenamiento y funciones recursivas.

Se incluyen ejercicios interactivos, representaciones gráficas y modificaciones de código para comprender las diferencias entre algoritmos y desarrollar soluciones más eficientes.

## Objetivos

- Comprender los fundamentos de la notación Big-O.
- Identificar la complejidad temporal de diferentes algoritmos.
- Comparar el crecimiento computacional mediante visualizaciones.

# **Parte B: Visualización del crecimiento**


Mueve `N` y observa cómo cambia el número aproximado de operaciones.  
No son tiempos reales, son modelos de crecimiento.


````python
def comparar_crecimientos(N=20):
    xs = list(range(1, N + 1))
    funciones = {
        "O(1)": [1 for n in xs],
        "O(log n)": [math.log2(n) for n in xs],
        "O(n)": xs,
        "O(n log n)": [n * math.log2(n) for n in xs],
        "O(n²)": [n**2 for n in xs],
    }

    plt.figure(figsize=(9, 5))
    for nombre, ys in funciones.items():
        plt.plot(xs, ys, label=nombre)
    plt.title("Comparación de crecimiento")
    plt.xlabel("N")
    plt.ylabel("Operaciones aproximadas")
    plt.legend()
    plt.grid(True)
    plt.show()

if WIDGETS_OK:
    interact(comparar_crecimientos, N=widgets.IntSlider(value=20, min=5, max=100, step=5))
else:
    comparar_crecimientos(20)
````

**Salida esperada:**

La gráfica muestra 5 curvas:

- O(1): permanece prácticamente horizontal.
- O(log n): crece muy lentamente.
- O(n): crece de manera proporcional a N.
- O(n log n): crece más rápido que O(n), pero más lento que O(n²).
- O(n²): crece rápidamente y termina siendo la curva más alta.

Se puede visualizar cómo los algoritmos con menor complejidad escalan mejor cuando aumenta el tamaño de los datos (N).

## **Reto 01:** Convierte de O(n²) a O(n)

El siguiente código revisa si hay repetidos, pero usa dos ciclos anidados.

**Tu tarea:** modifica la función para que use un `set` y quede en **O(n)**.


````python
# Modifica esta función

def tiene_repetidos(lista):
    for i in range(len(lista)):
        for j in range(i + 1, len(lista)):
            if lista[i] == lista[j]:
                return True
    return False

# Pruebas
print(tiene_repetidos([1, 2, 3, 4]))      # False
print(tiene_repetidos([1, 2, 3, 2]))      # True
````

# codigo modificado
### complejidad O(n²)

````python
# Función optimizada de O(n²) a O(n)

def tiene_repetidos(lista):
    vistos = set()

    for elemento in lista:
        if elemento in vistos:
            return True
        vistos.add(elemento)

    return False

# Pruebas
print(tiene_repetidos([1, 2, 3, 4]))      # False
print(tiene_repetidos([1, 2, 3, 2]))      # True
````

## **Reto 02:** Convierte de O(2ⁿ) a O(n)

El Fibonacci recursivo ingenuo recalcula muchas veces los mismos valores.

**Tu tarea:** modifica el código usando programación dinámica o una versión iterativa para que quede en **O(n)**.


````python
# Modifica esta función

def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n-1) + fibonacci(n-2)

# Pruebas
for i in range(8):
    print(i, fibonacci(i))
````

#codigo modificado

La complejidad ahora es:

O(n)

Porque solo se usa un ciclo que avanza desde 2 hasta n.

Antes era:O(2ⁿ)
porque la función recursiva llamaba muchas veces a los mismos valores, por ejemplo fibonacci(5), fibonacci(4), fibonacci(3), etc., repitiendo cálculos innecesarios.

````python
# Función optimizada de O(2ⁿ) a O(n)

def fibonacci(n):
    if n <= 1:
        return n

    anterior = 0
    actual = 1

    for i in range(2, n + 1):
        siguiente = anterior + actual
        anterior = actual
        actual = siguiente

    return actual

# Pruebas
for i in range(8):
    print(i, fibonacci(i))
````

## **Resumen:**

| Código / situación | Complejidad típica |
|---|---|
| Acceso directo | O(1) |
| Dividir el problema a la mitad | O(log n) |
| Un recorrido completo | O(n) |
| Ordenamientos eficientes | O(n log n) |
| Dos ciclos anidados sobre la misma entrada | O(n²) |
| Probar muchas combinaciones / recursión ramificada | O(2ⁿ) |
