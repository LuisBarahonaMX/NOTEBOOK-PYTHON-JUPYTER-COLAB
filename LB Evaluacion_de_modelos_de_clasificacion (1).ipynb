# Introducción

La evaluación de los modelos para la clasificación de datos no debe ser tomada a la ligera. Existen varias métricas que pueden usarse para ese fin. La más natural es la **exactitud** (porcentaje de casos correctamente predichos por el modelo); sin embargo, esa medida puede ser engañosa cuando las clases están desbalanceadas.

En aplicaciones de ciberseguridad, es común que la gran mayoría de los casos sean casos *seguros* mientras que un pequeño porcentaje puede corresponder a casos *fraudulentos*. Supón, por ejemplo, que una de cada mil transacciones en un sitio web es fraudulenta. Un modelo que clasifique a todas las transacciones como *segura* tendrá una exactitud del $99.9\%$, pero resultará completamente inútil.     

Es necesario revisar entonces otras métricas. En este documento, además de la exactitud, exploraremos la precisión, la exhaustividad, la sensibilidad y la especificidad.

# Matriz de confusión

La **matriz de confusión** es la base para la definición de las métricas para la evaluación de modelos de clasificación. Esta matriz, permite **contrastar** los resultados del modelo contra los reales. Considera un problema de clasificación binaria. La matriz de confusión tendrá la siguiente estructura:

\begin{matrix}
    \hline
    \text{Modelo/Realidad} & P & N \\
    \hline
    P & TP & FP \\
    N & FN & TN\\
    \hline
\end{matrix}

Siendo:

* $TP$: verdaderos positivos (true positive). Total de casos **correctamente** clasificados como positivos.
* $TN$: verdaderos negative (true negative). Total de casos **correctamente** clasificados como negativos.
* $FP$: falsos positivos (false positive). Total de casos **incorrectamente** clasificados como positivos.
* $FN$: falsos negativos (false negative). Total de casos **incorrectamente** clasificados como negativos.



# Exactitud [Accuracy]

$$Acc = \frac{TP+TN}{TP+TN+FP+FN}$$


# Precisión y exhaustividad [precision & recall ]


Definimos la **precisión** como:

$$Pre = \frac{TP}{TP+FP}$$

Para interpretar la precisión consideramos la proporción de casos positivos correctamente predichos [del total de casos predichos como positivos].

Mientras que la **exhaustividad** se define como:

$$Rec = \frac{TP}{TP+FN}$$

Interpretamos la exhaustividad como la fracción de casos positivos que son identificados por el clasificador [del total de casos positivos].

Dependiendo del contexto, puede ser mejor maximizar alguna de las dos medidas e <em>ignorar</em> a la otra. Algunos ejemplos son los siguientes:

1. **Buscador web**. Una vez que un usuario ingresa una consulta en un buscador web, éste selecciona a un conjunto de resultados que considera relevantes. Considera que puede haber miles o millones de documentos relacionados con la consulta del usuario, sin embargo, el buscador web buscará maximizar la <em>precisión</em> seleccionando solo a un conjunto limitado de ellos [por ejemplo 10] y esperando que la mayoría de ellos sean de verdad relevantes. Observa que en este contexto es preferible  sacrificar la exhaustividad [el que muchos documentos relevantes no sean mostrados al usuario no es mayor problema]
2. **Seguridad informática**. Para cuestiones de seguridad, es fundamental que un clasificador detecte a todos los casos <em>problemáticos</em>. Para ello se busca maximizar la <em>exhaustividad</em>. Observa que en este contexto, es preferible sacrificar la precisión [clasificar como problemáticos a algunos casos que no lo son puede provocar molestias a los usuarios del sistema informático pero es preferible a no detectar casos fraudulentos].

**Observa** que las ds medidas son complementarias.

# Sensibilidad y especificidad

Otras alternativas para evaluar la calidad de un modelo de clasificación binario son la **sensibilidad** y la **especificidad** definidas de la siguiente manera:

$$\text{Sensibilidad} = \frac{TP}{TP+FN}$$

Observa que la definción coincide con la de la exhaustividad.

La especificidad es:

$$\text{Especificidad} = \frac{TN}{TN+FP}$$

## Ejemplo

Para explorar los conceptos de precisión y exhaustividad, considera los datos del Titanic a los que puedes acceder mediante el siguiente código

```r
dat <- read.csv("https://raw.githubusercontent.com/jbryer/CompStats/master/Data/titanic3.csv")
```

```r
str(dat)
```

# Clasificador de referencia

Un clasificador que indique que ningún pasajero sobrevivió tendría una exactitud del 62%:  

```r
summary(dat$survived)
```

```r
1-mean(dat$survived)
```

Cualquier modelo que busquemos implementar debe tener una exactitud mayor a dicho valor.

```r
dat$pclass_f <- factor(dat$pclass, levels=c("1","2","3"))
summary(dat$pclass_f)
```

Separamos los datos en conjuntos de entrenamiento y prueba

```r
set.seed(123)
n <- nrow(dat)
n_train <- as.integer(0.7*n) #70% de la observaciones al conjunto de entrenamiento
indx <- sample(1:n,n_train)
train <- dat[indx,]
test <- dat[-indx,]
```

Ajustamos un clasificador por medio de regresión logística

```r
reg <- glm(survived~pclass_f+sex+age, data=train, family="binomial")
summary(reg)
```

En la siguientes celdas, construímos una matriz de confusión

```r
#predicción (recordar que la regresión logística modela una probabilidad)
p <- predict(reg, test, type="response")
p[1:5]
```

```r
#binarizamos
p_class <- ifelse(p > 0.5, "S", "N")
#p_class <- as.integer(p_class)

#comparamos lo predicho contra los valores reales
table(p_class, test$survived)
```

```r
#Algunas medidas de la calidad del ajuste

ex <- (97+153)/(97+153+48+23)

#precisión
pre <- 97/(97+23)

#exhaustividad y sensibilidad
rec <- 97/(97+48)

#especificidad
esp <- 153/(153+23)

print(paste0("exactitud: ",ex))
print(paste0("precisión: ",pre))
print(paste0("exhaustividad: ",rec))
print(paste0("especificidad: ",esp))
```

Observa que las medidas de calidad pueden variar si cambiamos el valor del umbral para definir las clases

```r
#binarizamos
p_class <- ifelse(p > 0.2, "S", "N")

#comparamos lo predicho contra los valores reales
table(p_class, test$survived)
```

```r
#Algunas medidas de la calidad del ajuste

#exactitud
ex <- (97+124)/(97+124+21+79)

#precisión
pre <- 124/(124+79)

#exhaustividad y sensibilidad
rec <- 124/(124+21)

#especificidad
esp <- 97/(97+79)

print(paste0("exactitud: ",ex))
print(paste0("precisión: ",pre))
print(paste0("exhaustividad/sensibilidad: ",rec))
print(paste0("especificidad: ",esp))
```

```r
#binarizamos
p_class <- ifelse(p > 0.8, "S", "N")

#comparamos lo predicho contra los valores reales
table(p_class, test$survived)
```

```r
#Algunas medidas de la calidad del ajuste

#exactitud
ex <- (176+45)/(176+45+100)

#precisión
pre <- 45/(45+0)

#exhaustividad y sensibilidad
rec <- 45/(45+100)

#especificidad
esp <- 176/(176)

print(paste0("exactitud: ",ex))
print(paste0("precisión: ",pre))
print(paste0("exhaustividad/sensibilidad: ",rec))
print(paste0("especificidad: ",esp))
```
